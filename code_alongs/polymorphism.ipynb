## Vectors

```python
class Vector:
    def __init__(self, *numbers: float):
        for number in numbers:
            if not isinstance(number, (float, int)):
                raise TypeError(f"number must be float or int not {type(number)}")

        if not numbers:
            raise ValueError("Vector can't be empty")

        self._numbers = tuple(float(number) for number in numbers)

    # read only numbers
    @property
    def numbers(self) -> tuple:
        return self._numbers

    # repr
    def __repr__(self):
        return f"Vector{self._numbers}"

    # math dunders
    def __add__(self, other):
        numbers =  (a+b for a,b in zip(self._numbers, other._numbers))
        if len(self) == len(other):
            return Vector(*numbers)
        else:
            raise TypeError("Missmatching dimensions")

    def __len__(self):
        '''returns number of elements in vector'''
        return len(self._numbers)
    
    def __abs__(self):
        '''returns a euclideam norm of a vector'''
        return sum(a**2 for a in self._numbers)**0.5

    def __getitem__(self, index: int):
        return self._numbers[index]

v1 = Vector(1, 2, 5)
v2 = Vector(1, -2, 3)
v3 = Vector(2, -3)

#######################
try:
    Vector("hi")
except TypeError as e:
    print(e)
try:
    Vector()
except ValueError as e:
    print(e)
try:
    v1 + v3
except TypeError as e:
    print(e)
#######################

print(len(v1))
print(v1 + v2)
print(v1[0])

abs(v1)
```
