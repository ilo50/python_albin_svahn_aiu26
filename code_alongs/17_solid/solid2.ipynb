# follows SOLID

```python
import json
from abc import ABC, abstractmethod

class Employee:
    def __init__(self, name, phone, salary, tax_rate):
        self.name = name
        self.phone = phone
        self.salary = salary
        self.tax_rate = tax_rate

    @property
    def net_salary(self):
        return self.salary * (1 - self.tax_rate)

class Report(ABC):
    @abstractmethod
    def content(self, employee: Employee) -> str: 
        ...

class NotWorkingreport(Report):
    ...


try:
    NotWorkingreport()
except TypeError as e:
    print(e)


class Excelreport(Report):
    def content(self, employee):
        print("excel content")

class txtreport(Report):
    def content(self, employee) -> str:
        return json.dumps({
            "Name": employee.name,
            "Name": employee.phone,
            "Salary": employee.salary,
            "Net Salary": employee.net_salary,
        },indent=3)

class JsonReport(Report):
    def content(self, employee: Employee) -> str:
        return json.dumps({
        "name": employee.name,
        "phone": employee.phone,
        "salary": employee.salary,
        "net_salary": employee.net_salary
        },indent=3)

def export(text: str, path: str):
    with open(path, "w") as file:
        file.write(text)

e1 = Employee("Anna", "041-487345", 37640, .25)
e1_report = txtreport().content(e1)
e1_json = JsonReport().content(e1)

export(e1_report, "anna.txt")
export(e1_json, "anna.json")
```
