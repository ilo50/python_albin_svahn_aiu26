## 1. Counting letters (*)

Let the user input a word:

a) print out the number of letters in the word. (*)

b) print out the number of uppercase and lowercase letters of the word (*)

````python
word = input("input word")

length = len(word)
print(f"Number of letters in your word: {length}")

count = 0
for char in word:
    if char.isupper():
        count += 1

print(f"Number of uppercase letters: {count}")
print(f"Number of lowercase letters: {length - count}")
````

## 2. Palindrome (*)

A palindrome is a sequence of characters that is the same, when read forward as backwards (ignoring spaces). For example:

- "Anna" is a palindrome
- "Ni talar bra latin"
- bjkjb

Let the user input a sequence of characters and check if it is a palindrome. (*)

````python
word = input("Input any string")
word_list = []
for char in word:
    if char != " ":
        word_list.append(char.lower())

reversed_list = word_list.copy()
reversed_list.reverse()

if (reversed_list == word_list):

    print(f"{word} is a palindrome")
else:
    print(f"{word} is not a palindrome")
````

## 3. Vowels (**)

Count the number of vowels in this sentence: "Pure mathematics is, in its way, the poetry of logical ideas"

````python
sentence = "Pure mathematics is, in its way, the poetry of logical ideas"

vowels = ("a", "e", "i", "o", "u", "y")

count = 0

for letter in sentence:

    if letter in vowels:

        count+=1

print(count)
````

## 4. Encryption (**)

Let the user input a word and:

a) encrypt the message by replacing each letter with the next letter. If the letter is in the end of the alphabet, use the first letter instead.

e.g. in Swedish: "höst" → "iatu"

````python
alphabet = "abcdefghijklmnopqrstuvwxyz"

word = input("Input a word")

encrypted_word = ""

for letter in word:

    encrypted_word += str(alphabet[alphabet.index(letter) + 1])

print(encrypted_word)
````

## 7. Tram (*)

Kokchun is a clumpsy computer user that doesn't take trams too often. Write a program to prompt the user for:

- number of times he/she wants to take tram in one month
- cost for one ticket
- cost for monthly card

The program should calculate if it's worth for him to buy monthly card or not. Make the program user friendly with clear error messages and ask again in case of input errors.

````python
nr_of_trips = int(input("How many times do you take the tram in one month?"))
ticket_cost = int(input("How much does a ticket cost in kr?"))
monthly_cost = int(input("How much does a monthly card cost in kr?"))

diff = nr_of_trips*ticket_cost - monthly_cost

if diff < 0:
    print(f"Bying single tickets are {abs(diff)} kr cheaper")
elif diff > 0:
    print(f"Monthly card is {(diff)} kr cheaper")
else:
    print("Either option is equally expensive")
````

### 8b) Sort the dice rolls from a) and write them to a separate row in the same textfile. (*)

### 8c) Count the number of fours in the dice rolls and write them to a separate row in the same textfile. (*)

````python
import random



with open("dice_rolls.txt", "w") as f:
    f.write("Simulate 20 dice rolls and write them to your textfile. \n")
    rolls = []
    for n in range(20):
        roll = random.randint(1, 6)
        f.write(f"{str(roll)} ")

        rolls.append(int(roll))

    f.write("\n")

    f.write("Sort the dice rolls from a and write them to a separate row in the same textfile. \n")
    rolls.sort()
    for r in rolls:
        f.write(f"{str(r)} ")

    f.write("\n")

    f.write("Count the number of fours in the dice rolls and write them to a separate row in the same textfile. \n")

    f.write(str(rolls.count(4)))
        
````
