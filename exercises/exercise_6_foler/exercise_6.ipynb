## 0. Unit conversion (*)

Create a class for converting US units to the metric system. It should have the following bound methods:

```python
class Unit_conversion:

    def __init__ (self, value = None):
        if not isinstance(value, (int, float)):
            print("incorrect type")
            self.__value = None
        else:
            self.__value = value

    def set_value(self, value = None):
        if not isinstance(value, (int, float)):
            print("incorrect type")
            self.__value = None
        else:
            self.__value = value
        
    def get_value(self):
        if self.__value is None:
            print("missing value")
        if isinstance(self.__value, (int, float)):
            return self.__value 
        else:
            print("incorrect type")
        
    def inch_to_cm(self):
        if self.__value is not None and isinstance(self.__value,(int, float)):
            self.__value = self.__value*2.54
        
    def foot_to_meters(self):
        if self.__value is not None and isinstance(self.__value,(int, float)):
            self.__value = self.__value*0.3048
        
    def pound_to_kg(self):
        if self.__value is not None and isinstance(self.__value,(int, float)):
            self.__value = self.__value*0.45359237
       
    def __repr__(self):
        return (f"value: {self.__value}")

inches = Unit_conversion("d")
feet = Unit_conversion(6)
pound = Unit_conversion(3)

inches.inch_to_cm()
feet.foot_to_meters()
pound.pound_to_kg()

print(inches)
print(feet)
print(pound)
```

# 3. Plotting library (**)

Create a class called PlotAnnotator which takes in a figure and axes instances and exposes a method called arrow, which serves the purpose of creating an annotation to a plot. Note that you might need other methods and/or properties, e.g. something that can return the figure instance.

Also add some settings so that you can reuse it for other plots. For simplicity you could have the settings directly in PlotAnnotator class, or if you'd like you could make use of inheritance or composition to separate concerns and make the classes have single responsibilities.

For example a manual test could look like this:

```python
import numpy as np
import matplotlib.pyplot as plt

class PlotAnnotator:

    def __init__(self, fig, ax):

        self.fig = fig
        self.ax = ax

    def arrow(self, text, cords1, cords2):
        
        self.ax.annotate(text, xy = cords1, xytext = cords2, arrowprops = dict(arrowstyle = "->"))
        self.return_figure()

    def return_figure(self):
        return self.fig, self.ax

fig, ax = plt.subplots()
np.random.seed(42)
x = np.linspace(0, 10, 100)
y = 2 * x + 5 + np.random.normal(0, 2, size=len(x))

ax.plot(x, y)

annotate = PlotAnnotator(fig, ax)
annotate.arrow("Big dip here", (7.5, 14.8), (8, 10))




```
