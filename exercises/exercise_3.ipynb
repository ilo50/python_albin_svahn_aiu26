# 4. Dice rolls convergence (**)

Simulate:

a) 100 dice rolls and count the number of outcome six. (*)

c) Use matplotlib to plot this list. (*)

```python
count = 0
for n in range (100):

    dice = random.randrange(1, 7)
    if dice == 6:
        count+=1

print(count)
```

b) 10, 100, 1000, 10000, 100000, 1000000 dice rolls. Count the number of outcome six in each simulation and store it in a list. Compute the probability of outcome six in each simulation. (**)

```python
list = []
roll_amounts = [10, 100, 1000, 10000, 100000, 1000000]
sixes = []

for amount in roll_amounts:
    count = 0
    for n in range(amount):

        dice = random.randint(1, 6)
        if dice == 6:
            count+=1

    list.append(count)

print(list)

for _ in range(6):
    val = (list[_] / roll_amounts[_])
    print(f"Distribution of six in simulation {_}: {round(val, 2)}")
    sixes.append(val)

```

c) Use matplotlib to plot this list. (*)

```python
x = ["10", "10^2", "10^3", "10^4", "10^5", "10^6"]
y = sixes

mplot.plot(x,y)
```

# 5. Monte Carlo simulation (***)

a) Simulate 5000 uniformly distributed points between -1 and 1, X ∼ U ( − 1 , 1 ) . Now calculate the Euclidean distances between each point and the origin (0,0), d i = x i 2 + y i 2 . For d i < 1 plot the point with one color, and for d i > 1 plot the point in another color, i = 1 , … , 5000 . (**)

```python
import math

x_scatter_red = []
x_scatter_blue = []

y_scatter_red = []
y_scatter_blue = []

for n in range(5000):

    x = random.uniform(-1, 1)
    y = random.uniform(-1, 1)

    d = (math.sqrt(x**2 + y**2))

    if d > 1:
        x_scatter_blue.append(x)
        y_scatter_blue.append(y)
    else:
        x_scatter_red.append(x)
        y_scatter_red.append(y)

mplot.scatter(x_scatter_red, y_scatter_red, color = "red", s=5)

mplot.scatter(x_scatter_blue, y_scatter_blue, color = "blue", s=5)

mplot.show
```

b) Calculate the fraction between number of inner points and number of outer points. What do you this will converge to when number of points reaches infinity? (**)

```python
tot = len(x_scatter_blue) + len(x_scatter_red)

frac_inner = len(x_scatter_red) / tot
frac_outer = len(x_scatter_blue) / tot

print(f"inner: {frac_inner}, outer: {frac_outer}")

# area of circle divided by are of square

circle = 1*math.pi
square = 2*2

print(f"Fraction of red points should converge to: {circle / square}")
```

# 6. A cute rabbit among two ferocious snakes (***)

There are three identical doors, behind two of them, there are ferocious snakes and behind one of them is a cute little rabbit. Your aim is to pick a door and get the cute rabbit. You start picking one door randomly and suddenly another door opens and snake slithers away pssss. Now you have a choice to either stay with your door or change to the remaining door.

a) What is your choice and why? (**)

- change doors, instead of 1/3 odds we know have 2/3 odds

b) Randomly assign which door has the rabbit, and randomly pick an initial door. Now simulate 10, 100, 1000, 10000, 100000, 1000000 times for staying and for switching doors. Calculate the proportion of finding the rabbits for each set of simulations. Plot the results (**)

```python
doors = ["snake", "snake", "snake"]

i = random.randint(0,2)
doors[i] = "rabbit"

sims = [10, 100, 1000, 10000, 100000, 1000000]

print("")
print("Sim with staying on the same door")
print("")

for sim in sims:
    count = 0
    for _ in range(sim):

        pick = random.randint(0,2)
        if (doors[pick] == "rabbit"):
            count+=1

    print(f"For sim with {sim} runs, proportion of correct guess is: {count/sim}")

print("")
print("Sim with switching the door")
print("")

for sim in sims:
    count = 0
    for _ in range(sim):

        pick = random.randint(0,2)

        snake_pick = random.randint(0,2)

        while doors[snake_pick] != "snake" or snake_pick == pick:
            snake_pick = random.randint(0,2)
            # one snake removed

        pick_2 = random.randint(0,2)

        while pick_2 == pick or pick_2 == snake_pick:

            pick_2 = random.randint(0,2)

        if doors[pick_2] == "rabbit":
            count += 1

    print(f"For sim with {sim} runs, proportion of correct guess is: {count/sim}")
```
