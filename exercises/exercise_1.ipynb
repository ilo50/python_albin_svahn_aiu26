## 2. Classification accuracy (*)

A machine learning model has been trained to detect fire. Here is the result of its predictions:

**True Positive (TP)**
- Reality: fire
- Predicted: fire
- Number of TP: 2

**False Positive (FP)**
- Reality: no fire
- Predicted: fire
- Number of FP: 2

**False Negative (FN)**
- Reality: fire
- Predicted: no fire
- Number of FN: 11

**True Negative (TN)**
- Reality: no fire
- Predicted: no fire
- Number of TN: 985

Calculate the accuracy using the following formula:

accuracy = (TP + TN) / (TP + TN + FP + FN)

Is this a good model? Why? Why not?

````python
acc = (2 + 985)/(2 + 985 + 2 + 11)

acc = acc*100

print(f"{acc}% accuracy")

# not really a good model since it has several false negatives, which is bad in this case
````
