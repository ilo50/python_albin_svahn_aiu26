## 0. Unit conversion (*)

Create a class for converting US units to the metric system. It should have the following bound methods:

```python
class Unit_conversion:

    def __init__ (self, value = None):
        if not isinstance(value, (int, float)):
            print("incorrect type")
            self.__value = None
        else:
            self.__value = value

    def set_value(self, value = None):
        if not isinstance(value, (int, float)):
            print("incorrect type")
            self.__value = None
        else:
            self.__value = value
        
    def get_value(self):
        if self.__value is None:
            print("missing value")
        if isinstance(self.__value, (int, float)):
            return self.__value 
        else:
            print("incorrect type")
        
    def inch_to_cm(self):
        if self.__value is not None and isinstance(self.__value,(int, float)):
            self.__value = self.__value*2.54
        
    def foot_to_meters(self):
        if self.__value is not None and isinstance(self.__value,(int, float)):
            self.__value = self.__value*0.3048
        
    def pound_to_kg(self):
        if self.__value is not None and isinstance(self.__value,(int, float)):
            self.__value = self.__value*0.45359237
       
    def __repr__(self):
        return (f"value: {self.__value}")

inches = Unit_conversion("d")
feet = Unit_conversion(6)
pound = Unit_conversion(3)

inches.inch_to_cm()
feet.foot_to_meters()
pound.pound_to_kg()

print(inches)
print(feet)
print(pound)
```

# 1. Person (*)

Create a class named Person, with parameterized constructor with the following parameters:

    name
    age
    email

Turn name, age, email into properties with following validations in their setters:

    name - must be string
    age - must be number between 0 and 125
    email - must include an @ sign

It should also have __repr__ method to represent the Person class in a neat way.

Also create a method say_hello() that prints

Hi, my name is ..., I am ... years old, my email address is ...


```python
class Person:

    def __init__(self):
        self.name = None
        self.age = None
        self.email = None

    def set_name(self, name):
        if isinstance(name, str):
            self.name = name
        else:
            print(f"{name} is not a valid name")

    def set_age(self, age):
        if isinstance(age, int) and 0 < age < 125 :
            self.age = age
        else:
            print(f"{age} is not a valid age")

    def set_email(self, email):
        if isinstance(email, str) and "@" in email:
            self.email = email
        else:
            print(f"{email} is not a valid email")

    def __repr__(self):
        if (self.name and self.age and self.email):
            return(f"Name: {self.name} Age: {self.age} Email: {self.email}")
        else:
            return("person is missing attributes, cannot print")

    def say_hello(self):
        print(f"Hi, my name is {self.name}, I am {self.age} years old, my email address is {self.email}")


person1 = Person()
person1.set_name("anna")
person1.set_age(21)
person1.set_email("bla@blabla.com")

    
print(person1)

person1.say_hello()

    
```
