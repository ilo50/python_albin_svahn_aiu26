## 3. Divisible (*)

Let the user input a number. Check if the number is

a)   even or odd
b)   is divisible by 5
c)   is divisble by 5 and odd


```python
num = int(input("Enter a number"))

even = False
div5 = False
both = False

if num % 2 == 0:
    even = True
if num % 5 == 0:
    div5 = True
if num % 2 != 0 and num % 5 == 0:
    both = True

print(f"Number is even? {even}, Number is divisible bby 5? {div5}, Number is both odd and divisible with 5? {both}")
```

## 4. Luggage size (*)

The maximum allowed luggage size for boarding an airplane is:

    weight: 8kg
    dimensions: 55x40x23cm (length x width x height)

Let the user input weight, length, width and height of the luggage. The program should check if the luggage is allowed or not.

```python
weight = int(input("Input weight in kg"))
length = int(input("input length in cm"))
width = int(input("input width in cm"))
height = int(input("input height in cm"))

if weight <= 8 and length <= 55 and width <= 40 and height <= 23:
    print("Your baggage is allowed")
else:
    print("Your baggage is not allowed")
    
```

## 6. Multiplication game (*)

```python
import random

score = 0

while(True):

    x = random.randrange(1, 10 + 1)
    y = random.randrange(1, 10 + 1)

    answer = int(input(f"What is {x} times {y}?"))
    
    if answer == x*y:
        score += 1
        print(f"You are correct!, your score is now: {score}")

        a = input("Do you want to play again? y / n")
        if a == "n":
            print(f"Thank you for playing!")
            break
```

## 9. Check convergence (**)

Use a while statement to compute the following sums:

a)   sum = 1 + 1 2 + 1 4 + 1 8 + ⋯ + 1 2 n

b)   sum = 1 − 1 3 + 1 5 − 1 7 + ⋯ + ( − 1 ) n 2 n + 1

Try different values on n to see which value it converges to.

```python
n = 100
sum = 0
i = 0

while i != n:

    sum+=1/2**i
    i+=1

print(sum)
```

```python
n = 100
sum = 0
i = 0

while i != n:

    top = (-1)**n
    bot = 2*n + 1

    sum+= top/bot
    i+=1

print(sum)
```
