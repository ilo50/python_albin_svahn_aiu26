## 1. Euclidean distance (\*)

The formula for Euclidean distance in 2D between $P : (p_1, p_2)$ and $Q : (q_1, q_2)$ is:

$$d(P, Q) = \sqrt{(p_1 - q_1)^2 + (p_2 - q_2)^2}$$

**a)** Create a function that takes two points as input parameters and return the Euclidean distance between them. (\*)

````python
import math

def euc(p, q):

    return round(math.sqrt((p[0] - q[0])**2 + (p[1] - q[1])**2), 2)
````

**b)** Let the user input two points. Call the function using the user's input points. (\*)

````python
p = input("Input P as: p1, p2")
q = input("Input Q as: q1, q2")

p = p.split(",")
q = q.split(",")

p1 = int(p[0].strip())
p2 = int(p[1].strip())

q1 = int(q[0].strip())
q2 = int(q[1].strip())

P = (p1, p2)
Q = (q1, q2) 

print(euc(P, Q))
````

**c)** Use your function to calculate distances between the origin (0, 0) and each of these points: (10, 3), (-1, -9), (10, -10), (4, -2), (9, -10). (\*)

````python
points = [(10, 3), (-1, -9), (10, -10), (4, -2), (9, -10)]
origin = (0, 0)

for point in points:
    print(euc(origin, point))
````

Create a function that takes a name as an input and:

- removes all leading and trailing blank spaces
- makes the first character of each name capitalized, and the rest lowercase

Use your function on this list of strings:


## 4. Change (\*\*)

Create a function that takes a value as input parameter and prints out the banknotes and coins in Swedish currency representing this value. For example, 5289 would give the following printout:

```
5st 1000-lapp
1st 200-lapp
1st 50-lapp
1st 20-lapp
1st 10-krona
1st 5-krona
2st 2-krona
```

Now let the user input a value, and use the function to calculate the change.

````python
def change(amount):

    count = [0, 0, 0, 0, 0, 0, 0, 0, 0]
    notes = [1000, 500, 200, 50, 20, 10, 5, 2, 1]

    for index, note in enumerate(notes):

        if amount//note != 0:

            count[index] = amount//note
            amount = amount - note*(amount//note)

    for i in range(len(count)):

        c = count[i]
        if c != 0:
            if notes[i] < 20:
                print(f"{c}st {notes[i]}-krona")
            else:
                print(f"{c}st {notes[i]}-lapp")
            
    
change(5289)
````

## 6. Dice simulation (\*)

Simulate 1000000 dice rolls and save the number of ones, twos, ..., sixes in a dictionary. Then print them out in the terminal. (\*)

````python
import random

counts = [0, 0, 0, 0, 0, 0]

for n in range(1000000):

    dice = random.randint(1,6)
    
    counts[dice-1] += 1

dices = {
  "ones": counts[0],
  "twos": counts[1],
  "threes": counts[2],
  "fours": counts[3],
  "fives": counts[4],
  "sixes": counts[5]
}

print(dices)
````
